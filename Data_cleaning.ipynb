```python
import pandas as pd
```

```python
# 1. LOAD DATA: Load the raw dataset safely. 
# We explicitly tell pandas to read CID and TID as text ('str') so we don't lose digit precision.
df = pd.read_csv('Dataset.csv',dtype = {'TID':str,'CID':str})
```

```python
df.head(5)
```

```python
# 2. FIX MISSING VALUES: 
# Replace the 27,585 'NaN' values in 'Discount Name' with the label 'No Discount'
df['Discount Name'] = df['Discount Name'].fillna('No Discount')
```

```python
# 3. FIX NEGATIVE REVENUE (The Anomaly):
# Create a "mask" (a filter) that finds all rows where the discount is larger than the gross price
over_discount_mask = df['Discount Amount (INR)'] > df['Gross Amount']
```

```python
# Apply the cap: Set the discount amount equal to the gross amount for those specific rows
df.loc[over_discount_mask, 'Discount Amount (INR)'] = df.loc[over_discount_mask, 'Gross Amount']
```

```python
# Set the final Net Amount to 0.0 for those specific rows
df.loc[over_discount_mask, 'Net Amount'] = 0.0
```

```python
# 4. FIX DATES: Convert the text string '30/08/2023 20:27:08' into a smart Datetime object.
# This allows us to extract month, day, and time later in Stage 2.
df['Purchase Date'] = pd.to_datetime(df['Purchase Date'], format='%d/%m/%Y %H:%M:%S')
```

```python
# 5. VALIDATE THE CLEANING:
print("--- CLEANING VALIDATION ---")
print("Remaining Missing 'Discount Name':", df['Discount Name'].isnull().sum())
print("Remaining Negative 'Net Amount':", (df['Net Amount'] < 0).sum())
print("Data Type of 'Purchase Date':", df['Purchase Date'].dtype)
```

```python
# 6. EXPORT CLEAN DATA: Save the cleaned data to a new file for the next stages.
df.to_excel('cleaned_dataset_perfect.xlsx', index=False, engine='openpyxl')
print("Success! Cleaned data saved natively as 'cleaned_dataset_perfect.xlsx'")
```
