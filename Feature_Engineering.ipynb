```python
import pandas as pd
import numpy as np
```

```python
# 1. LOAD CLEANED EXCEL DATA
# We use parse_dates to ensure pandas reads the dates correctly
df = pd.read_excel('cleaned_dataset_perfect.xlsx', dtype={'TID': str, 'CID': str}, parse_dates=['Purchase Date'])    
```

```python
# 2. FEATURE ENGINEERING: TIME DIMENSIONS
# Extracting month, day of week, and hour for trend analysis
df['Year'] = df['Purchase Date'].dt.year
df['Month'] = df['Purchase Date'].dt.month_name()
df['Day of Week'] = df['Purchase Date'].dt.day_name()
df['Hour'] = df['Purchase Date'].dt.hour
```

```python
# 3. FEATURE ENGINEERING: PAYMENT GROUPING
def categorize_payment(method):
    if pd.isna(method):
        return 'Unknown'
    elif 'UPI' in method:
        return 'UPI'
    elif 'Card' in method or 'Net Banking' in method:
        return 'Cards & Banking'
    else:
        return 'Cash on Delivery'

df['Payment_Category'] = df['Purchase Method'].apply(categorize_payment)
```

```python
# 4. FEATURE ENGINEERING: DISCOUNT DEPTH
df['Discount_Percentage'] = np.where(df['Gross Amount'] > 0, 
                                     (df['Discount Amount (INR)'] / df['Gross Amount']) * 100, 
                                     0)
```

```python
# 5. PREVIEW AND SAVE
print("--- NEW FEATURES PREVIEW ---")
print(df[['Purchase Date', 'Year', 'Month', 'Day_of_Week', 'Hour', 'Payment_Category', 'Discount_Percentage']].head())

# Save this enriched dataset as our final analytical file
df.to_excel('engineered_dataset.xlsx', index=False, engine='openpyxl')
print("\nSuccess! Engineered data saved natively as 'engineered_dataset.xlsx'")
```
